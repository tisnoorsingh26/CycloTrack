```python
import imdtrack as imd

bt = imd.load().clean(fix_dates=True)
df = bt.observations

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

display(df.head())
```

```python

# Sort observations in chronological order within each storm
df = df.sort_values(["storm_id", "time"]).copy()

# Time difference between consecutive observations of each storm
df["hours_since_previous"] = (
    df.groupby("storm_id")["time"].diff().dt.total_seconds() / 3600
)

print("Unique storms:", df["storm_id"].nunique())
print("\nObservations by basin:")
print(df["basin"].value_counts())

print("\nTime interval distribution (hours):")
print(df["hours_since_previous"].value_counts().sort_index())

print("\nObservations per storm:")
print(df.groupby("storm_id").size().describe())

```

```python

import numpy as np
import pandas as pd
from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import StandardScaler

# Keep ocean cyclones with valid positions and timestamps
data = df[
    df["basin"].isin(["BOB", "ARB"])
    & df["lat"].between(-40, 40)
    & df["lon"].between(30, 130)
].dropna(subset=["storm_id", "time", "lat", "lon"]).copy()

data = data.sort_values(["storm_id", "time"])

# Remove duplicate timestamps within each storm
data = data.drop_duplicates(["storm_id", "time"], keep="first")

# Features: position, recorded wind, and time of year
data["wind"] = data["wind"].fillna(data.groupby("storm_id")["wind"].transform("median"))
data["wind"] = data["wind"].fillna(data["wind"].median())
data["hour_sin"] = np.sin(2 * np.pi * data["time"].dt.hour / 24)
data["hour_cos"] = np.cos(2 * np.pi * data["time"].dt.hour / 24)
data["month_sin"] = np.sin(2 * np.pi * data["time"].dt.month / 12)
data["month_cos"] = np.cos(2 * np.pi * data["time"].dt.month / 12)

feature_cols = ["lat", "lon", "wind", "hour_sin", "hour_cos",
                "month_sin", "month_cos"]

# Make windows using exact timestamps; don't bridge large gaps.
X, Y, groups = [], [], []
input_hours = [0, 3, 6, 9, 12, 15, 18, 21]

for storm_id, g in data.groupby("storm_id"):
    g = g.sort_values("time").set_index("time")

    # Require an uninterrupted 3-hourly input sequence.
    for start in g.index:
        input_times = [start + pd.Timedelta(hours=h) for h in input_hours]
        target24 = start + pd.Timedelta(hours=45)
        target48 = start + pd.Timedelta(hours=69)

        # The final input is at +21h; targets are +45h and +69h.
        if not all(t in g.index for t in input_times):
            continue
        if target24 not in g.index or target48 not in g.index:
            continue

        x = g.loc[input_times, feature_cols].to_numpy(dtype=np.float32)
        y = g.loc[[target24, target48], ["lat", "lon"]].to_numpy(dtype=np.float32)

        X.append(x)
        Y.append(y)
        groups.append(storm_id)

X = np.asarray(X, dtype=np.float32)
Y = np.asarray(Y, dtype=np.float32)
groups = np.asarray(groups)

print("Input sequences:", X.shape)
print("Targets:", Y.shape)
print("Unique storms represented:", len(np.unique(groups)))

```

```python

from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import StandardScaler
import numpy as np

# First split: 80% of storms for development, 20% held out for testing
splitter = GroupShuffleSplit(n_splits=1, test_size=0.20, random_state=42)
trainval_idx, test_idx = next(splitter.split(X, Y, groups=groups))

# Second split: reserve some training storms for validation
splitter_val = GroupShuffleSplit(n_splits=1, test_size=0.20, random_state=43)
train_rel, val_rel = next(
    splitter_val.split(X[trainval_idx], Y[trainval_idx], groups=groups[trainval_idx])
)

train_idx = trainval_idx[train_rel]
val_idx = trainval_idx[val_rel]

# Normalize using training data only
x_scaler = StandardScaler()
x_scaler.fit(X[train_idx].reshape(-1, X.shape[-1]))

def scale_x(arr):
    return x_scaler.transform(
        arr.reshape(-1, arr.shape[-1])
    ).reshape(arr.shape).astype(np.float32)

# Scale target coordinates using training target positions only
y_scaler = StandardScaler()
y_scaler.fit(Y[train_idx].reshape(-1, 2))

X_train = scale_x(X[train_idx])
X_val = scale_x(X[val_idx])
X_test = scale_x(X[test_idx])

Y_train = y_scaler.transform(Y[train_idx].reshape(-1, 2)).reshape(-1, 2, 2)
Y_val = y_scaler.transform(Y[val_idx].reshape(-1, 2)).reshape(-1, 2, 2)
Y_test = y_scaler.transform(Y[test_idx].reshape(-1, 2)).reshape(-1, 2, 2)

print("Training sequences:", X_train.shape[0])
print("Validation sequences:", X_val.shape[0])
print("Test sequences:", X_test.shape[0])

print("Training storms:", len(np.unique(groups[train_idx])))
print("Validation storms:", len(np.unique(groups[val_idx])))
print("Test storms:", len(np.unique(groups[test_idx])))
```

```python

import tensorflow as tf
from tensorflow.keras import layers, models, callbacks

tf.keras.utils.set_random_seed(42)

model = models.Sequential([
    layers.Input(shape=(X_train.shape[1], X_train.shape[2])),
    layers.LSTM(64, return_sequences=False),
    layers.Dense(32, activation="relu"),
    layers.Dense(4),
    layers.Reshape((2, 2))
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="mse",
    metrics=["mae"]
)

early_stop = callbacks.EarlyStopping(
    monitor="val_loss",
    patience=12,
    restore_best_weights=True
)

history = model.fit(
    X_train, Y_train,
    validation_data=(X_val, Y_val),
    epochs=100,
    batch_size=32,
    callbacks=[early_stop],
    verbose=1
)

model.summary()
```

```python

import numpy as np

# Predict on the held-out test storms
pred_scaled = model.predict(X_test, verbose=0)

# Convert predictions and targets back to latitude/longitude
pred = y_scaler.inverse_transform(
    pred_scaled.reshape(-1, 2)
).reshape(-1, 2, 2)

actual = y_scaler.inverse_transform(
    Y_test.reshape(-1, 2)
).reshape(-1, 2, 2)

# Great-circle distance in kilometres
def haversine_km(a, b):
    lat1, lon1 = np.radians(a[..., 0]), np.radians(a[..., 1])
    lat2, lon2 = np.radians(b[..., 0]), np.radians(b[..., 1])

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    h = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
    )
    h = np.clip(h, 0, 1)
    return 6371.0 * 2 * np.arcsin(np.sqrt(h))

# Baseline: continue the last observed 3-hour movement
raw_test = X[test_idx]
last = raw_test[:, -1, :2]       # last input position
previous = raw_test[:, -2, :2]  # position 3 hours earlier
movement_3h = last - previous

baseline = np.stack([
    last + movement_3h * 8,   # 24 hours ahead
    last + movement_3h * 16   # 48 hours ahead
], axis=1)

# Compare errors
for i, horizon in enumerate([24, 48]):
    lstm_errors = haversine_km(pred[:, i, :], actual[:, i, :])
    base_errors = haversine_km(baseline[:, i, :], actual[:, i, :])

    print(f"\n{horizon}-hour forecast")
    print(f"LSTM mean error:       {lstm_errors.mean():.1f} km")
    print(f"LSTM median error:     {np.median(lstm_errors):.1f} km")
    print(f"Baseline mean error:   {base_errors.mean():.1f} km")
    print(f"Baseline median error: {np.median(base_errors):.1f} km")
```

```python

import tensorflow as tf
from tensorflow.keras import layers, models, callbacks, regularizers

tf.keras.utils.set_random_seed(42)

improved_model = models.Sequential([
    layers.Input(shape=(X_train.shape[1], X_train.shape[2])),
    layers.LSTM(
        16,
        dropout=0.25,
        recurrent_dropout=0.0,
        kernel_regularizer=regularizers.l2(0.01)
    ),
    layers.Dense(
        16,
        activation="relu",
        kernel_regularizer=regularizers.l2(0.01)
    ),
    layers.Dropout(0.25),
    layers.Dense(4),
    layers.Reshape((2, 2))
])

improved_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0005),
    loss="mse",
    metrics=["mae"]
)

stop = callbacks.EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True
)

reduce_lr = callbacks.ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=4,
    min_lr=0.00001
)

history2 = improved_model.fit(
    X_train, Y_train,
    validation_data=(X_val, Y_val),
    epochs=80,
    batch_size=32,
    callbacks=[stop, reduce_lr],
    verbose=1
)
```

```python

pred2_scaled = improved_model.predict(X_test, verbose=0)

pred2 = y_scaler.inverse_transform(
    pred2_scaled.reshape(-1, 2)
).reshape(-1, 2, 2)

for i, horizon in enumerate([24, 48]):
    errors = haversine_km(pred2[:, i, :], actual[:, i, :])

    print(
        f"{horizon}-hour forecast: "
        f"mean error = {errors.mean():.1f} km, "
        f"median error = {np.median(errors):.1f} km"
    )
```

```python
import numpy as np
import pandas as pd

# Work on a fresh copy of the filtered data
motion_data = data.copy()
motion_data = motion_data.sort_values(["storm_id", "time"])

# Changes from the previous observation in each storm
motion_data["delta_lat"] = (
    motion_data.groupby("storm_id")["lat"].diff()
)
motion_data["delta_lon"] = (
    motion_data.groupby("storm_id")["lon"].diff()
)

# Time elapsed since the previous observation, in hours
motion_data["delta_hours"] = (
    motion_data.groupby("storm_id")["time"]
    .diff()
    .dt.total_seconds() / 3600
)

# Approximate north/south and east/west movement in km/hour
motion_data["north_kmh"] = (
    motion_data["delta_lat"] * 111.0
    / motion_data["delta_hours"]
)

motion_data["east_kmh"] = (
    motion_data["delta_lon"]
    * 111.0
    * np.cos(np.radians(motion_data["lat"]))
    / motion_data["delta_hours"]
)

# Remove undefined or invalid movement measurements
motion_data[["north_kmh", "east_kmh"]] = (
    motion_data[["north_kmh", "east_kmh"]]
    .replace([np.inf, -np.inf], np.nan)
)

print("Rows:", len(motion_data))
print(
    motion_data[
        ["delta_lat", "delta_lon", "north_kmh", "east_kmh"]
    ].describe().round(2)
)
```

```python

# Build sequences using position, wind, time, and motion features

feature_cols_v2 = [
    "lat", "lon", "wind",
    "hour_sin", "hour_cos",
    "month_sin", "month_cos",
    "north_kmh", "east_kmh",
    "delta_lat", "delta_lon"
]

X2, Y2, groups2 = [], [], []

input_hours = [0, 3, 6, 9, 12, 15, 18, 21]

for storm_id, g in motion_data.groupby("storm_id"):
    g = g.sort_values("time").set_index("time")

    # Require all input features to be present
    for start in g.index:
        input_times = [
            start + pd.Timedelta(hours=h)
            for h in input_hours
        ]

        target24 = start + pd.Timedelta(hours=45)
        target48 = start + pd.Timedelta(hours=69)

        if not all(t in g.index for t in input_times):
            continue

        if target24 not in g.index or target48 not in g.index:
            continue

        x = g.loc[input_times, feature_cols_v2]

        if x.isna().any().any():
            continue

        y = g.loc[
            [target24, target48], ["lat", "lon"]
        ].to_numpy(dtype=np.float32)

        X2.append(x.to_numpy(dtype=np.float32))
        Y2.append(y)
        groups2.append(storm_id)

X2 = np.asarray(X2, dtype=np.float32)
Y2 = np.asarray(Y2, dtype=np.float32)
groups2 = np.asarray(groups2)

print("X2 shape:", X2.shape)
print("Y2 shape:", Y2.shape)
print("Distinct storms:", len(np.unique(groups2)))
```

```python

from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import StandardScaler
import numpy as np

# Split by storm, not by individual sequence
splitter = GroupShuffleSplit(
    n_splits=1, test_size=0.20, random_state=42
)
trainval_idx2, test_idx2 = next(
    splitter.split(X2, Y2, groups=groups2)
)

splitter_val = GroupShuffleSplit(
    n_splits=1, test_size=0.20, random_state=43
)
train_rel2, val_rel2 = next(
    splitter_val.split(
        X2[trainval_idx2],
        Y2[trainval_idx2],
        groups=groups2[trainval_idx2]
    )
)

train_idx2 = trainval_idx2[train_rel2]
val_idx2 = trainval_idx2[val_rel2]

# Fit feature scaler using training data only
x_scaler2 = StandardScaler()
x_scaler2.fit(
    X2[train_idx2].reshape(-1, X2.shape[-1])
)

def scale_x2(arr):
    return x_scaler2.transform(
        arr.reshape(-1, arr.shape[-1])
    ).reshape(arr.shape).astype(np.float32)

X_train2 = scale_x2(X2[train_idx2])
X_val2 = scale_x2(X2[val_idx2])
X_test2 = scale_x2(X2[test_idx2])

# Scale target coordinates using training targets only
y_scaler2 = StandardScaler()
y_scaler2.fit(Y2[train_idx2].reshape(-1, 2))

def scale_y2(arr):
    return y_scaler2.transform(
        arr.reshape(-1, 2)
    ).reshape(-1, 2, 2).astype(np.float32)

Y_train2 = scale_y2(Y2[train_idx2])
Y_val2 = scale_y2(Y2[val_idx2])
Y_test2 = scale_y2(Y2[test_idx2])

print("Train:", X_train2.shape, "storms:",
      len(np.unique(groups2[train_idx2])))
print("Validation:", X_val2.shape, "storms:",
      len(np.unique(groups2[val_idx2])))
print("Test:", X_test2.shape, "storms:",
      len(np.unique(groups2[test_idx2])))
```

```python

import tensorflow as tf
from tensorflow.keras import layers, models, callbacks, regularizers

tf.keras.utils.set_random_seed(42)

model2 = models.Sequential([
    layers.Input(shape=(X_train2.shape[1], X_train2.shape[2])),
    layers.LSTM(
        24,
        dropout=0.2,
        kernel_regularizer=regularizers.l2(0.001)
    ),
    layers.Dense(
        16,
        activation="relu",
        kernel_regularizer=regularizers.l2(0.001)
    ),
    layers.Dropout(0.2),
    layers.Dense(4),
    layers.Reshape((2, 2))
])

model2.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0005),
    loss="mse",
    metrics=["mae"]
)

early_stop2 = callbacks.EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True
)

reduce_lr2 = callbacks.ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=4,
    min_lr=0.00001
)

history3 = model2.fit(
    X_train2, Y_train2,
    validation_data=(X_val2, Y_val2),
    epochs=80,
    batch_size=32,
    callbacks=[early_stop2, reduce_lr2],
    verbose=1
)
```

```python

pred2_scaled = model2.predict(X_test2, verbose=0)

pred2 = y_scaler2.inverse_transform(
    pred2_scaled.reshape(-1, 2)
).reshape(-1, 2, 2)

actual2 = y_scaler2.inverse_transform(
    Y_test2.reshape(-1, 2)
).reshape(-1, 2, 2)

# Simple baseline: continue the most recent 3-hour movement
raw_test2 = X2[test_idx2]

last_position = raw_test2[:, -1, :2]
previous_position = raw_test2[:, -2, :2]
movement_3h = last_position - previous_position

baseline2 = np.stack([
    last_position + movement_3h * 8,
    last_position + movement_3h * 16
], axis=1)

# Compare errors using great-circle distance
for i, horizon in enumerate([24, 48]):
    lstm_errors = haversine_km(
        pred2[:, i, :], actual2[:, i, :]
    )
    baseline_errors = haversine_km(
        baseline2[:, i, :], actual2[:, i, :]
    )

    print(f"\n{horizon}-hour forecast")
    print(
        f"LSTM:     mean = {lstm_errors.mean():.1f} km, "
        f"median = {np.median(lstm_errors):.1f} km"
    )
    print(
        f"Baseline: mean = {baseline_errors.mean():.1f} km, "
        f"median = {np.median(baseline_errors):.1f} km"
    )
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models, regularizers
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from sklearn.preprocessing import StandardScaler

# Inputs: 8 historical points with movement features
# Target: change from the last observed position to +45 hours
# The last input point is at +21h, so +45h means 24 hours ahead.

last_position = X2[:, -1, :2]  # latitude, longitude
future_24h = Y2[:, 0, :]       # first target = 24h ahead

# Predict displacement, not absolute coordinates
Y24 = future_24h - last_position

# Use the same held-out-storm split as before
X_train24 = X2[train_idx2]
X_val24   = X2[val_idx2]
X_test24  = X2[test_idx2]

Y_train24 = Y24[train_idx2]
Y_val24   = Y24[val_idx2]
Y_test24  = Y24[test_idx2]

# Scale inputs using training data only
x_scaler24 = StandardScaler()
x_scaler24.fit(X_train24.reshape(-1, X_train24.shape[-1]))

def scale_X24(x):
    return x_scaler24.transform(
        x.reshape(-1, x.shape[-1])
    ).reshape(x.shape)

X_train24_s = scale_X24(X_train24)
X_val24_s   = scale_X24(X_val24)
X_test24_s  = scale_X24(X_test24)

# Scale displacement targets using training data only
y_scaler24 = StandardScaler()
Y_train24_s = y_scaler24.fit_transform(Y_train24)
Y_val24_s   = y_scaler24.transform(Y_val24)
Y_test24_s  = y_scaler24.transform(Y_test24)

tf.keras.utils.set_random_seed(42)

model24 = models.Sequential([
    layers.Input(shape=(X_train24_s.shape[1], X_train24_s.shape[2])),
    layers.LSTM(
        32,
        dropout=0.15,
        kernel_regularizer=regularizers.l2(0.0005)
    ),
    layers.Dense(16, activation="relu"),
    layers.Dense(2)
])

model24.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0005),
    loss="mse",
    metrics=["mae"]
)

history24 = model24.fit(
    X_train24_s, Y_train24_s,
    validation_data=(X_val24_s, Y_val24_s),
    epochs=100,
    batch_size=32,
    callbacks=[
        EarlyStopping(
            monitor="val_loss",
            patience=12,
            restore_best_weights=True
        ),
        ReduceLROnPlateau(
            monitor="val_loss",
            patience=5,
            factor=0.5,
            min_lr=1e-6
        )
    ],
    verbose=1
)
```

```python

import os
import joblib
import zipfile
from IPython.display import FileLink, display

# Folder to store both forecasting models and their scalers
save_dir = "/kaggle/working/cyclone_models"
os.makedirs(save_dir, exist_ok=True)

# Save 24-hour model and scalers
model24.save(os.path.join(save_dir, "cyclone_24h.keras"))
joblib.dump(x_scaler24, os.path.join(save_dir, "x_scaler_24h.pkl"))
joblib.dump(y_scaler24, os.path.join(save_dir, "y_scaler_24h.pkl"))

# Save 48-hour model and scalers
model48.save(os.path.join(save_dir, "cyclone_48h.keras"))
joblib.dump(x_scaler48, os.path.join(save_dir, "x_scaler_48h.pkl"))
joblib.dump(y_scaler48, os.path.join(save_dir, "y_scaler_48h.pkl"))

# Create a ZIP containing all six files
zip_path = "/kaggle/working/cyclone_models.zip"

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as z:
    for filename in os.listdir(save_dir):
        full_path = os.path.join(save_dir, filename)
        if os.path.isfile(full_path):
            z.write(full_path, arcname=filename)

# Verify the saved files
print("Saved files:")
with zipfile.ZipFile(zip_path, "r") as z:
    for filename in z.namelist():
        print(f"  {filename}")

print("\nZIP size:", round(os.path.getsize(zip_path) / (1024 * 1024), 2), "MB")
print("\nClick the link below to download:")
display(FileLink(zip_path))
```
